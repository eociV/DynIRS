Settings: GPU T4 x2, Internet on. Inputs: dynirs-code, dynirs-dv, (dynirs-omega), (previous output). Edit QUEUE / ONLY, Run all, then Save Version (Save & Run All) to keep /kaggle/working.

```python
QUEUE = 'configs/queues/dyn/t4_v8n_s0_c.json'
ONLY = ''  # e.g. 'dyn_v8n_C_s0'
import os, subprocess, sys
omega = '/kaggle/input/dynirs-omega' if os.path.exists('/kaggle/input/dynirs-omega') else None
prev = [p for p in os.listdir('/kaggle/input') if p not in ('dynirs-code', 'dynirs-dv', 'dynirs-omega')]
args = [sys.executable, '/kaggle/input/dynirs-code/scripts/kaggle_session.py', '--queue', QUEUE,
        '--data-in', '/kaggle/input/dynirs-dv', '--work', '/kaggle/working', '--code-in', '/kaggle/input/dynirs-code']
args += ['--omega-in', omega] if omega else []
args += ['--prev-out', '/kaggle/input/' + prev[0]] if prev else []
args += ['--only', *ONLY.split()] if ONLY else []
print(subprocess.run(args).returncode)
```
